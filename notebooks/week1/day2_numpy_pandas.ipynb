Build a small NumPy array and inspect basic array operations.

```python
import numpy as np

a = np.array([[1, 2], [3, 4]])
print("Array:\n", a)
print("Shape:", a.shape)
print("Mean:", a.mean())
print("Array multiplied by 2:\n", a * 2)
print("Transpose:\n", a.T)
```

Create a random RGB image and inspect its dimensions, first pixel, and range.

```python
image = np.random.randint(0, 256, (224, 224, 3))
print("Image shape:", image.shape)
print("RGB values at pixel [0, 0]:", image[0, 0])
print("Minimum pixel value:", image.min())
print("Maximum pixel value:", image.max())
```

Create a file-label DataFrame, count its labels, and select fake-labeled files.

```python
import pandas as pd

files = pd.DataFrame({
    "file_name": ["video_01.mp4", "video_02.mp4", "video_03.mp4", "video_04.mp4", "video_05.mp4", "video_06.mp4"],
    "label": ["real", "fake", "real", "fake", "fake", "real"],
})
print("Label counts:\n", files["label"].value_counts())
print("\nFake-labeled files:\n", files[files["label"] == "fake"])
```

Convert text labels to numeric IDs and print the counts for each class.

```python
label_to_id = {"real": 0, "fake": 1}
files["label_id"] = files["label"].map(label_to_id)
print("Rows with numeric labels:\n", files)
print("\nClass counts by label ID:\n", files["label_id"].value_counts().sort_index())
```
