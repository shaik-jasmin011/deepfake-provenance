Load the handwritten-digits dataset and prepare its features and labels.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

digits = load_digits()
X = digits.data
y = digits.target
images = digits.images
```

Display the first 10 handwritten digits in a 2-by-5 image grid.

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 5))
for index, axis in enumerate(axes.flat):
    axis.imshow(images[index], cmap="gray")
    axis.set_title(f"Label: {y[index]}")
    axis.axis("off")
fig.tight_layout()
plt.show()
```

Plot the number of examples in each digit class.

```python
class_counts = np.bincount(y)
plt.figure(figsize=(8, 4))
plt.bar(np.arange(len(class_counts)), class_counts)
plt.xlabel("Digit class")
plt.ylabel("Number of images")
plt.title("Class distribution in the digits dataset")
plt.xticks(np.arange(len(class_counts)))
plt.show()
```

Print the dataset dimensions, number of classes, and pixel-value range.

```python
print("Dataset shape:", X.shape)
print("Number of classes:", len(np.unique(y)))
print("Pixel value range:", (images.min(), images.max()))
```
